###난수 기초와 생성

```python
import numpy as np

rg = np.random.default_rng(1)
rg.random() # [0,1] 실수 난수
```

```python
rg.random(4)
```

```python
rg.random( (3, 4, 2) )
```

```python
import numpy as np

rg = np.random.default_rng(12)
rg.integers(1, 7, (3, 4) ) # [1,7] 정수 난수
```

```python
rg.integers(low=1, high=10, size=(2, 8)) #[1,10]
```

```python
import numpy as np

rg = np.random.default_rng(12)
rg.standard_normal(10) # 표준 정규 분포 실수 난수
```

```python
rg.standard_normal( (3, 4) )
```

```python
rg.standard_normal( (10, 5) )
```
