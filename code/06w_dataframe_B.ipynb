```python
from IPython.display import display_html
def display_side_by_side(*args):
    """여러 데이터프레임 비교가 쉽게 옆쪽으로 표시한다"""
    html_str=''
    for df in args:
        html_str += df.to_html() + ' '*4
    display_html(html_str.replace('table','table style="display:inline"'), raw=True)
```

```python
display_side_by_side(df, df.cumsum(axis=1))
```

```python
display_side_by_side(df, df.cumsum(axis=0))
```

```python
df.mean()
```

```python
df.mean(axis=1)
```

```python
df.mean(axis=None)
```

```python
df.max()
```

```python
display_side_by_side(df,df.mean(axis=1).to_frame())
```

```python
display_side_by_side(df,df.max(axis=1).to_frame())
```

```python
display_side_by_side(df.max().to_frame(),df.min().to_frame())
```

```python
display_side_by_side(df.max(axis=0).to_frame(),df.min(axis=0).to_frame())
```
